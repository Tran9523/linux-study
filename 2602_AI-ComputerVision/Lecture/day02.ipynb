```python
# 반복문
# 제어문 중 하나
# while, for
# while : 대개 반복횟수가 정해져 있지 않을 때
# for : 대개 반복횟수가 정해져 있을 때

# while 조건식:
#     실행문장
# Ctrl + C / Ctrl + V로 코드를 붙여넣었다?
# -> 규칙이 있다 -> 반복문을 사용!!
# 반복횟수를 모르겠다 -> while

# 사용자가 양의 정수를 입력하면
# 누적하여 합한 결과를 출력!!
# 단, 사용자가 -1을 입력하면 종료!!

total = 0
num = int(input("양의 정수 입력 >>"))
while num != -1:
    total += num # 누적합을 저장하는 변수
    print(f"누적 힙 : {total}")
    num = int(input("양의 정수 입력 >>"))
```

```python
# 탐색(Search)

# 이진 탐색(Binary Search)
list31 = [1, 7, 16, 25, 30, 33, 41, 66, 78, 90]
search = 78
# 없는 조건을 넣는다면 무한 반복문 logic에 빠지게 됨
# 그렇게 되면 low가 high보다 높아지는 현상이 나타나게 됨
low_index = 0
high_index = len(list31)-1
mid_index = (low_index + high_index // 2)
# 찾고자 하는 값이 없을 때 변경
is_find = False
while low_index <= high_index: #탐색값이 없는 상황을 예외
    if list31[mid_index] > search:
        high_index = mid_index - 1
    elif list31[mid_index] < search:
        low_index = mid_index + 1
    else :
        print(f"찾고자 하는 값은 index : {mid_index} 위치에 있습니다.")
        is_find = True
        break
    mid_index = (low_index + high_index) // 2

if not is_find:
    print("해당 값은 존재하지 않습니다.")
```

```python
# 변수 : 반복되는 값을
# 반복문 : 반복되는 실행문장
# 함수* : 반복되는 Logic

# C에서는 함수
# 선언, 정의, 호출 

# Python에서 함수
# 정의, 호출

# 정의부
# def 함수명(매개변수):
#     실행문장
#     return 반환값
# 내 렙과 펫 렙을 합쳐서 선물 주는 logic
def gift(user_level, pet_level):
    total_gift = (user_level + pet_level) * 2 - 1
    return total_gift

# 호출부
# 함수명(전달인자)
print(gift(1, 1))
print(gift(3, 5))
print(gift(5, 7))
```

```python
# 오늘의 함수 예제 목표 : 완전수 구하기!!
# 완전수(Perfect Number)
# 자기 자신을 제외한 약수의 합이
# 자기 자신과 같은 수
# ex. 6 -> 1 2 3 6 : 첫 번째 완전수
# ex. 8 -> 1 2 4 8 : 완전수가 아님
```

```python
# 1) is_divisor(num1, num2) -> bool
# 만약, num2가 num1의 약수라면 True
# 약수가 아니라면 False 반환하는 함수 정의
def is_divisor(num1, num2):
    return num1 % num2 == 0

print(is_divisor(10,3))
```

```python
# 2) 약수를 나열!!
# print_divisors(num)
def print_divisors(num):
    print(f"{num}의 약수 :", end=" ")
    for i in range(1, num//2 +1):
        if is_divisor(num, i):
            print(f"{i}", end=" ")
    print(num)

print_divisors(10)
print_divisors(28)
print_divisors(60)
# 내가 정의한 다른 함수를 호출해서 사용 가능
# 재귀함수라고 자기자신을 호출할 수 도 있음
```

```python
# 3) get_sum_of_divisors(num)
# 약수의 총합을 구하는 함수
def get_sum_of_divisors(num):
    total = 0
    # num을 바로 넣으면 의미 희석됨
    for i in range(1, num//2 +1):
        if is_divisor(num, i):
            total += i
    total += num
    return total
    
print(get_sum_of_divisors(6))
print(get_sum_of_divisors(28))
```

```python
# 4) is_perfect_number(num)

def is_perfect_number(num):
    return get_sum_of_divisors(num) == num *2

print(is_perfect_number(10))
print(is_perfect_number(6))
print(is_perfect_number(28))
```

```python
# 5) 구하기
from tqdm import tqdm
value1 = int(input("시작 범위 >>"))
value2 = int(input("끝 범위 >>"))
n = 1
for i in tqdm(range(value1, value2)):
    if is_perfect_number(i):
        print(f"{n}번째 완전수 : {i}")
        n += 1
# 첫번째 완전수 : 6
# 두번째 완전수 : 28
# 세번째 완전수 : 100 ~ 1000
# 네번째 완전수 : 1000 ~ 10_000
```

```python
# 함수의 결과값은 항상 하나다.

def cal(num1, num2):
    return num1 + num2, num1 - num2

print(cal(10, 5))
# 결과값 2개 아닌가요?
# (15, 5) -> 튜플(tuple)이다.
```

```python
def add_numbers(*nums): # 가변 매개변수
    total = 0
    for num in nums:
        total += num
    return total
# 값이 튜플로 넘어옴
print(add_numbers(10, 5))
print(add_numbers(10, 5, 3))
```
