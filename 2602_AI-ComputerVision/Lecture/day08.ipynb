```python
# Qt UI를 만드는 라이브러리
# C++ Qt -> 객체 지향 프로그래밍

from PyQt5.QtWidgets import (QApplication, QWidget, QPushButton, QLabel, QLineEdit)
import sys # 해당 App을 우리 윈도우에 띄워보자!!

# 객체 지향 프로그래밍
# 상속 : 검증된 클래스 사용 가능
# 예) 포유류 <- 코끼리 (상속 화살표는 자식에서 부모로 가도록)
class MyApp(QWidget):
    def __init__(self):
        # 부모 안에 있는 멤버 가져다 쓰자!!
        super().__init__() # 부모 호출
        self.initUI()

    def initUI(self):
        self.setWindowTitle("First App")
        # 위치 사이즈 -> setGeometry(x, y, width, height)
        self.setGeometry(1200, 700, 400, 200) # 2모니터면 -로도 가능
        
        self.btn = QPushButton("눌러봐", self)
        self.btn.move(100, 80)
        self.btn.resize(self.btn.sizeHint()) # 버튼 크기 알아서 결정
        # 버튼이 눌러졌을 때 (이벤트가 발생했다!!)
        # 버튼 이벤트 : 버튼 누름, 버튼 뗌, 눌렀다 뗌(Click), 상태가 바뀜
        self.btn.clicked.connect(self.setBGColor)

        # 새로운 버튼 만들어서 배경 색깔 핫핑크
        btn2 = QPushButton("버튼2", self)
        btn2.move(100, 120)
        # btn2는 외부에서 바라볼 일이 없음
        # lambda expression : 함수 생성과 동시에 호출 -> 버림
        # btn2.clicked.connect(lambda : self.setStyleSheet("background-color:hotpink"))
        btn2.pressed.connect(lambda : self.setStyleSheet("background-color:red"))
        btn2.released.connect(lambda : self.setStyleSheet("background-color:blue"))

        # 텍스트(+a) 출력하는 QLabel
        # 배치... 키보드 (나중에 마우스)
        self.label = QLabel("로그인중", self)
        self.label.move(40, 40)

        self.edit_id = QLineEdit("", self)
        self.edit_id.setPlaceholderText("ID를 입력하세요")
        self.edit_pw = QLineEdit("", self)
        self.edit_pw.setPlaceholderText("PW를 입력하세요")

        # pw editor에서 엔터 치면 self.login() 호출
        self.edit_pw.returnPressed.connect(self.login)

        self.edit_id.move(240, 60)
        self.edit_pw.move(240, 100)

        self.btn_login = QPushButton("로그인", self)
        self.btn_login.move(280, 140)

        # 1) 함수 login 만들어서
        # edit_id 와 edit_pw에 있는 문자열 가져오기
        # 미리 지정한 id와 pw면 성공 로직
        # 아니면 실패 로직

        # 2) btn_login과 login() 연결!!
        self.btn_login.clicked.connect(self.login)

        # 추가) 로그인 실패 시 self.label 을 "로그인 실패"
        # 로그인 성공 시 edit_id, edit_pw 내용을 clear

        self.show()

    def login(self):
        id = self.edit_id.text()
        pw = self.edit_pw.text()
        if id == "sereu" and pw == "1234":
            self.label.setText(f"{id}님 환영합니다.")
        else:
            self.label.setText("로그인 실패")
        self.label.resize(self.label.sizeHint())
        self.edit_id.clear()
        self.edit_pw.clear()




    def setBGColor(self): # 배경컬러 - 함수명 자유
        # btn이 눌러졌을 때, 실행되는 부분!!
        # self -> 가장 바깥 요소
        self.setStyleSheet("background-color:#00FF00")
        # 버튼의 문구 수정!! -> btn이 전역 변수라 self.btn 선언 먼저
        self.btn.setText("눌렀네?")

# flask server 만들 때!!
# 여러분들이 python을 사용하는 가장 큰 이유? : 모듈화
# .py -> 자유롭게 import 가능
# python 인터프리터 언어
# -> 우리가 import를 진행하면 해당 py파일이 전부 로드
# 개발자가(배포자가) 실행하길 원하지 않는 부분 -> __name__ 활용

# 내가 직접 run, import를 통해 run
# 내가 직접 run -> __name__ 이 __main__으로 바뀜
# __name__ -> day08, __main__ -> day08
# 외부에서 exam.py에서 import day08을 했다
# __name__ -> exam, __main__ -> day08

# 외부에서 import 했을 때, 실행되어야 되지 않는
# 크리티컬한 코드들은 __name__으로 처리하자

if __name__ == "__main__": # 이 파트는 고정
    # 만약, 윈도우 창이 떠 있으면 이전 창 사용!!
    app = QApplication.instance() # 창이 없으면 None, 있으면 그 정보
    if app is None:
        app = QApplication(sys.argv)
    exam = MyApp() # 지금 쓰진 않지만 미리 넣어두기
    sys.exit(app.exec_()) # 윈도우 창이 꺼지면 app도 꺼버리자
```

```python
from PyQt5.QtWidgets import (QApplication, QWidget, QPushButton, QLabel, QLineEdit,
                             QVBoxLayout, QHBoxLayout)
from PyQt5.QtGui import QPixmap
import sys

img_list = os.listdir("imgs/")

class MyApp(QWidget):
    def __init__(self):
        super().__init__()
        self.initUI()

    def initUI(self):
        self.resize(400, 500)
        vbox = QVBoxLayout()
        self.label = QLabel("")
        # self.label에 이미지 로드해보자!!
        self.item = 0
        self.update_image()
        self.label.setFixedSize(400, 400)
        self.label.setScaledContents(True)

        hbox = QHBoxLayout()
        btn_prev = QPushButton("<")
        btn_next = QPushButton(">")

        # 문제)
        # btn_next를 누르면 img_list의 다음 index 이미지 로드!!
        # btn_prev를 누르면 img_list의 이전 index 이미지 로드!!

        # 5번째 이미지(index 4)에서 btn_next를 누르면
        # 1번째 이미지(index 0)으로 이동
        # 1번째 이미지(index 0)에서 btn_prev를 누르면
        # 5번째 이미지(index 4)으로 이동

        # 1) 현재 어떤 index>>
        # 2) 무조건 self.label에 img_list를 사용중
        #    ㄴ 그러면 index만 주어지면 로드 되도록 함수
        # 3) 이걸 바탕으로 next & prev 도 연결하면 됨

        btn_prev.clicked.connect(self.prev)
        btn_next.clicked.connect(self.next)

        vbox.addWidget(self.label)
        vbox.addLayout(hbox)
        hbox.addWidget(btn_prev)
        hbox.addWidget(btn_next)
        self.setLayout(vbox)

        self.show()
    
    def update_image(self):
        img_path = f"imgs/{img_list[self.item]}"
        self.label.setPixmap(QPixmap(img_path))

    def prev(self):
        self.item = (self.item - 1) % len(img_list)
        self.update_image()

    def next(self):
        self.item = (self.item + 1) % len(img_list)
        self.update_image()
        


if __name__ == "__main__":
    app = QApplication.instance()
    if app is None:
        app = QApplication(sys.argv)
    exam = MyApp()
    sys.exit(app.exec_())
```

```python
from PyQt5.QtWidgets import (QApplication, QWidget, QPushButton, QLabel, QLineEdit,
                             QVBoxLayout, QHBoxLayout, QGridLayout)
import sys

class MyApp(QWidget):
    def __init__(self):
        super().__init__()
        self.initUI()

    def initUI(self):

        # 계산기 만들어보자
        vbox = QVBoxLayout()
        grid = QGridLayout()
        self.label = QLabel("0")

        vbox.addWidget(self.label)
        vbox.addSpacing(50)
        vbox.addLayout(grid)

        btn_names = [["7", "8", "9", "*"],
                     ["4", "5", "6", "-"],
                     ["1", "2", "3", "+"],
                     ["C", "0", "=", "/"]]
        
        self.result = ""

        # grid에 버튼 16개 만들어서 넣어보자
        for i in range(4):
            for j in range(4):
                btn = QPushButton("")
                btn.setText(btn_names[i][j])
                grid.addWidget(btn, i, j)
                btn.clicked.connect(self.cal)



        self.setLayout(vbox)
        self.resize(300,150)
        self.show()

    def cal(self):
        # 어떤 버튼에서 날 호출했는지!!
        btn = self.sender() # 특정 위젯에서 호출 시 해당 위젯 정보를 반환
        key = btn.text()
        
        # 1) 숫자, +, -, *, / 누르면
        # self.result에 해당 문자열 누적!!
        # self.label에 내용 갱신(self.result 로 갱신)

        # 2) C 누르면
        # self.result 비우고, self.label은 0으로

        # 3) = 누르면
        # evel() 통해 계산한 결과를 self.label에 출력

        if key == "C":
            self.result = ""
            self.label.setText("0")
        elif key == "=":
            try:
                value = eval(self.result) if self.result else 0
                self.result = str(value)
                self.label.setText(self.result)
            except :
                self.result = ""
                self.label.setText("Error")
        else:
            if self.result == "0":
                self.result = key
            else:
                self.result += key
            self.label.setText(self.result)


if __name__ == "__main__":
    app = QApplication.instance()
    if app is None:
        app = QApplication(sys.argv)
    exam = MyApp()
    sys.exit(app.exec_())
```

```python
from PyQt5.QtWidgets import (
    QApplication, QWidget, QPushButton, QLabel,
    QVBoxLayout, QGridLayout
)
from PyQt5.QtCore import Qt
import sys

class MyApp(QWidget):
    def __init__(self):
        super().__init__()
        self.initUI()

    def initUI(self):
        # 창 크기 고정 (디스플레이가 커지는 느낌 방지)
        self.setFixedSize(280, 300)
        self.setWindowTitle("Calculator")

        vbox = QVBoxLayout()
        grid = QGridLayout()

        # 디스플레이
        self.label = QLabel("0")
        self.label.setAlignment(Qt.AlignRight | Qt.AlignVCenter)
        self.label.setFixedHeight(70)
        self.label.setProperty("display", True)

        vbox.addWidget(self.label)
        vbox.addLayout(grid)

        btn_names = [
            ["7", "8", "9", "*"],
            ["4", "5", "6", "-"],
            ["1", "2", "3", "+"],
            ["C", "0", "=", "/"]
        ]

        self.result = ""

        # 버튼 생성 + 스타일 구분 property 부여
        for i in range(4):
            for j in range(4):
                text = btn_names[i][j]
                btn = QPushButton(text)
                btn.setFixedSize(60, 50)

                if text.isdigit():
                    btn.setProperty("kind", "num")
                elif text in "+-*/":
                    btn.setProperty("kind", "op")
                elif text == "C":
                    btn.setProperty("kind", "clear")
                elif text == "=":
                    btn.setProperty("kind", "eq")

                btn.clicked.connect(self.cal)
                grid.addWidget(btn, i, j)

        self.setLayout(vbox)

        # 스타일(QSS)
        self.setStyleSheet("""
            QWidget { background-color: #121212; }

            QLabel[display="true"]{
                background-color:#1e1e1e;
                color:white;
                font-size:32px;
                padding:12px;
                border-radius:12px;
            }

            QPushButton{
                border:none;
                border-radius:12px;
                font-size:18px;
                font-weight:bold;
            }

            QPushButton[kind="num"]{ background-color:#2a2a2a; color:white; }
            QPushButton[kind="num"]:pressed{ background-color:#3a3a3a; }

            QPushButton[kind="op"]{ background-color:#ff9f0a; color:black; }
            QPushButton[kind="op"]:pressed{ background-color:#ffb84d; }

            QPushButton[kind="clear"]{ background-color:#ff453a; color:white; }
            QPushButton[kind="clear"]:pressed{ background-color:#ff6b63; }

            QPushButton[kind="eq"]{ background-color:#0a84ff; color:white; }
            QPushButton[kind="eq"]:pressed{ background-color:#4aa3ff; }
        """)

        self.show()

    def format_display(self, value):
        """
        유효숫자 제한으로 표시 문자열 만들기
        - float: 유효숫자 10개로 제한
        - 너무 길면 뒤에 … 처리
        """
        if isinstance(value, float):
            s = f"{value:.10g}"   # 유효숫자 10개
        else:
            s = str(value)

        if len(s) > 12:
            s = s[:12] + "…"
        return s

    def cal(self):
        btn = self.sender()
        key = btn.text()

        if key == "C":
            self.result = ""
            self.label.setText("0")

        elif key == "=":
            try:
                value = eval(self.result) if self.result else 0
                self.result = str(value)  # 내부 계산값은 유지
                self.label.setText(self.format_display(value))  # 표시는 유효숫자 제한
            except:
                self.result = ""
                self.label.setText("Error")

        else:
            # 앞자리 0 방지: "0" 상태에서 숫자 누르면 교체
            if self.result == "0" and key.isdigit():
                self.result = key
            else:
                self.result += key

            self.label.setText(self.result if self.result else "0")


if __name__ == "__main__":
    app = QApplication.instance()
    if app is None:
        app = QApplication(sys.argv)
    exam = MyApp()
    sys.exit(app.exec_())
```

```python
from PyQt5.QtWidgets import (QApplication, QWidget, QPushButton, QLabel, QLineEdit,
                             QVBoxLayout, QHBoxLayout, QGridLayout)
import sys
import random
import time

class MyApp(QWidget):
    def __init__(self):
        super().__init__()
        self.initUI()

    def initUI(self):

        # 1) 기본 UI 틀 잡기
        vbox = QVBoxLayout()

        self.label = QLabel("1 to 9")
        self.btn_start = QPushButton("GAME START")
        grid = QGridLayout()

        vbox.addWidget(self.label)
        vbox.addWidget(self.btn_start)
        vbox.addLayout(grid)

        self.btn_list = []

        for i in range(3):
            for j in range(3):
                btn = QPushButton(" ")
                btn.setMinimumSize(100, 80)
                btn.setEnabled(False) # 시작 전 버튼 Push X
                self.btn_list.append(btn)
                grid.addWidget(btn, i, j)
                btn.clicked.connect(self.game)

        self.setLayout(vbox)
        self.btn_start.clicked.connect(self.start_game)
        
        self.show()

    # GAME START 버튼 눌렀을 때, 모든 버튼 활성화!!
    # 1 ~ 9 까지의 숫자를 각 btn.setText(str(숫자**))
    # 숫자 자체를 넣게 되면, 내부 로직이 돌아갈 우려 있음

    # 1) enumerate -> index를 활용
    # 2) temp의 요소를 btn에 추가
    #    ㄴ그 값을 zip으로 묶어서 돌리기
    # 3) pop()
    def start_game(self):
        nums = list(range(1, 10))
        random.shuffle(nums)
        for btn in self.btn_list:
            btn.setEnabled(True)
            btn.setText(str(nums.pop()))
        self.answer = 1 # 시작 번호로 같이 지정
        self.label.setText(f"1 to 9 start")
        self.start_time = time.perf_counter()



    def game(self):
        # 1, 2, 3 버튼을 눌렀을 때 호출되는 함수
        btn = self.sender()
        key = btn.text()
        # 순차적으로 1, 2, 3, 4 눌렀을 때
        # 정답을 눌렀을 때, 해당 버튼을 비활성화
        if int(key) == self.answer:
            btn.setText("")
            btn.setEnabled(False)
            self.answer += 1
        if self.answer == 10:
            self.end_time = time.perf_counter()
            self.label.setText(f"시간 : {self.end_time - self.start_time : .3f}초")

    # 만약 9를 눌렀으면
    # GAME START부터 9 누른 순간까지 시간을
    # self.label에 기입
    # 예) 5.04초

if __name__ == "__main__":
    app = QApplication.instance()
    if app is None:
        app = QApplication(sys.argv)
    exam = MyApp()
    sys.exit(app.exec_())
```

```python
from PyQt5.QtWidgets import (QApplication, QWidget, QPushButton, QLabel,
                             QVBoxLayout, QGridLayout)
from PyQt5.QtCore import Qt, QTimer
import sys
import random
import time

class MyApp(QWidget):
    def __init__(self):
        super().__init__()
        self.answer = 1
        self.start_time = 0.0
        self.is_running = False
        self.initUI()

    def initUI(self):
        self.setWindowTitle("1 to 9 GAME")
        self.setFixedSize(360, 420)

        vbox = QVBoxLayout()
        vbox.setContentsMargins(18, 18, 18, 18)
        vbox.setSpacing(14)

        self.label = QLabel("1 to 9")
        self.label.setAlignment(Qt.AlignCenter)
        self.label.setProperty("role", "title")

        self.btn_start = QPushButton("GAME START")
        self.btn_start.setFixedHeight(52)
        self.btn_start.setProperty("role", "start")

        grid = QGridLayout()
        grid.setSpacing(12)

        vbox.addWidget(self.label)
        vbox.addWidget(self.btn_start)
        vbox.addLayout(grid)
        vbox.addStretch(1)

        self.btn_list = []
        for i in range(3):
            for j in range(3):
                btn = QPushButton("")
                btn.setFixedSize(95, 75)
                btn.setEnabled(False)
                btn.setProperty("role", "cell")
                btn.clicked.connect(self.game)
                self.btn_list.append(btn)
                grid.addWidget(btn, i, j)

        self.setLayout(vbox)
        self.btn_start.clicked.connect(self.start_game)

        # ✅ 폰트/굵기(너무 두껍지 않게) + UI 스타일
        # - Windows 기본: Segoe UI (가독성 좋고 굵기 덜함)
        # - 한글도 필요하면 "맑은 고딕" 추가 가능
        self.setStyleSheet("""
            QWidget {
                background-color: #f3f3f3;
                font-family: "Segoe UI", "Malgun Gothic", "Arial";
                font-weight: 400;
            }

            QLabel[role="title"] {
                font-size: 32px;
                font-weight: 500;
                color: #111;
                padding: 10px;
            }

            QPushButton[role="start"] {
                background-color: white;
                border: 1px solid #cfcfcf;
                border-radius: 10px;
                font-size: 14px;
                font-weight: 500;
                padding: 12px;
            }
            QPushButton[role="start"]:pressed {
                background-color: #ececec;
            }

            QPushButton[role="cell"] {
                background-color: white;
                border: 1px solid #cfcfcf;
                border-radius: 10px;
                font-size: 20px;
                font-weight: 500;
                color: #111;
            }
            QPushButton[role="cell"]:pressed {
                background-color: #e9e9e9;
            }
            QPushButton[role="cell"]:disabled {
                background-color: #f7f7f7;
                border: 1px solid #e5e5e5;
                color: #c0c0c0;
            }

            /* 정답/오답 순간 피드백용 (동적으로 property를 바꿔서 적용) */
            QPushButton[role="cell"][state="ok"] {
                background-color: #d8f5d8;
                border: 1px solid #7acb7a;
            }
            QPushButton[role="cell"][state="bad"] {
                background-color: #ffd7d7;
                border: 1px solid #ff7a7a;
            }
        """)

        self.show()

    def start_game(self):
        nums = list(range(1, 10))
        random.shuffle(nums)

        for btn, n in zip(self.btn_list, nums):
            btn.setEnabled(True)
            btn.setText(str(n))
            btn.setProperty("state", "")
            btn.style().unpolish(btn)
            btn.style().polish(btn)

        self.answer = 1
        self.is_running = True
        self.label.setText("1 to 9 start")
        self.start_time = time.perf_counter()

    def flash_state(self, btn: QPushButton, state: str, ms: int = 200):
        """버튼을 잠깐 ok/bad 색으로 표시했다가 원복"""
        btn.setProperty("state", state)
        btn.style().unpolish(btn)
        btn.style().polish(btn)

        def reset():
            btn.setProperty("state", "")
            btn.style().unpolish(btn)
            btn.style().polish(btn)

        QTimer.singleShot(ms, reset)

    def end_game(self):
        self.is_running = False
        end_time = time.perf_counter()
        self.label.setText(f"{end_time - self.start_time:.2f} sec")

        # 게임 끝나면 남은 버튼도 다 비활성화
        for b in self.btn_list:
            b.setEnabled(False)

    def game(self):
        if not self.is_running:
            return

        btn = self.sender()
        key = btn.text().strip()
        if not key:
            return

        n = int(key)

        # ✅ 정답
        if n == self.answer:
            # 정답 피드백 (초록 반짝)
            self.flash_state(btn, "ok", 200)

            # 0.2초 후에 지우고 비활성화 (반짝 보이게)
            def clear_btn():
                btn.setText("")
                btn.setEnabled(False)

            QTimer.singleShot(200, clear_btn)

            # 마지막(9)면 종료
            if self.answer == 9:
                QTimer.singleShot(220, self.end_game)
                return

            self.answer += 1

        # ❌ 오답
        else:
            self.flash_state(btn, "bad", 200)

if __name__ == "__main__":
    app = QApplication.instance()
    if app is None:
        app = QApplication(sys.argv)
    exam = MyApp()
    sys.exit(app.exec_())
```

```python
from PyQt5.QtWidgets import (QApplication, QWidget, QPushButton, QLabel, QLineEdit,
                             QVBoxLayout, QHBoxLayout, QGridLayout, QListWidget)
import sys
import random

class MyApp(QWidget):
    def __init__(self):
        super().__init__()
        self.initUI()
    
    def initUI(self):
        self.resize(400, 600)

        vbox = QVBoxLayout()
        self.view = QListWidget()
        self.edit = QLineEdit()
        self.btn = QPushButton("게임 시작")
        vbox.addWidget(self.view)
        vbox.addWidget(self.edit)
        vbox.addWidget(self.btn)
        self.status = 0
        # 숫자 야구 게임
        # 랜덤한 3자리 숫자를 설정 (각 자리는 중복 X)
        # [0 ~ 9] shuffle 된 리스트에서 3개 팝! -> self.answer에 담기
        temp = list(range(10))
        random.shuffle(temp)
        digit_1 = temp.pop()
        digit_2 = temp.pop()
        digit_3 = temp.pop()
        self.answer = f"{digit_1}{digit_2}{digit_3}"

        self.btn.clicked.connect(self.confirm)
        self.edit.returnPressed.connect(self.confirm)

        # self.edit에 123 입력 후 btn 클릭하면
        # 해당 숫자가 자리까지 일치 -> Strike
        # 자리 일치 X, 숫자는 있다 -> Ball
        # 해당 숫자 X -> Out
        # 사용자에게 123 1S 1B 1OUT 알려준다 (QListWidget에 추가)
        # 3S 나오면 종료!! (축하합니다)
        self.setLayout(vbox)
        self.show()
    def confirm(self):
        if self.status == 0:
            temp = list(range(10))
            random.shuffle(temp)
            digit_1 = temp.pop()
            digit_2 = temp.pop()
            digit_3 = temp.pop()
            self.answer = f"{digit_1}{digit_2}{digit_3}"
            self.view.addItem("게임이 시작되었습니다.")
            self.status = 1
        else:
            guess = self.edit.text()
            if len(guess) != 3 or (not guess.isdigit()) or len(set(guess)) != 3:
                self.view.addItem("[입력 오류] 중복 없는 3자리 숫자를 입력하세요. 예) 123")
                return
            strike, ball, out = 0, 0, 0
            # 첫 번째 자리가 Strike, Ball, Out
            for i in range(3):
                if guess[i] == self.answer[i]:
                    strike += 1
                elif guess[i] in self.answer:
                    ball += 1
                else:
                    out += 1
            self.view.addItem(f"{guess} -> {strike}S {ball}B {out}OUT")

            if strike == 3:
                self.view.addItem("🎉 축하합니다! 정답입니다!")
                self.status = 0
if __name__ == "__main__":
    app = QApplication.instance()
    if app is None:
        app = QApplication(sys.argv)
    exam = MyApp()
    sys.exit(app.exec_())
```

```python
from PyQt5.QtWidgets import (QApplication, QWidget, QPushButton, QLabel, QLineEdit,
                             QVBoxLayout, QHBoxLayout, QGridLayout, QListWidget)
import sys
import random

class MyApp(QWidget):
    def __init__(self):
        super().__init__()
        self.initUI()

    def initUI(self):
        vbox = QVBoxLayout()

        self.view = QListWidget()
        self.edit = QLineEdit()
        self.edit.setPlaceholderText("서로 다른 3자리 숫자 입력 (예: 123)")
        self.btn = QPushButton("게임 시작")

        self.status = 0 # 0: 대기(게임 시작 전), 1: 진행(정답 입력)
        self.tries = 0
        self.answer = None

        self.btn.clicked.connect(self.new_game)
        self.edit.returnPressed.connect(self.new_game)

        vbox.addWidget(self.view)
        vbox.addWidget(self.edit)
        vbox.addWidget(self.btn)

        self.setLayout(vbox)
        self.resize(400, 600)
        self.setWindowTitle("숫자 야구")
        self.show()

    def start(self):
        nums = list(range(10))
        random.shuffle(nums)
        self.answer = nums[:3]   # 중복 없는 3자리
        self.status = 1
        self.tries = 0

        self.view.clear()
        self.view.addItem("게임 시작! 서로 다른 3자리 숫자를 맞혀보세요.")
        # 디버그용: 정답 확인
        # self.view.addItem(f"(DEBUG) 정답: {self.answer}")

        self.btn.setText("<<정답 입력>>")
        self.edit.clear()
        self.edit.setFocus()

    def validate_input(self, s: str):
        s = s.strip()
        if len(s) != 3:
            return None, "3자리로 입력하세요."
        if not s.isdigit():
            return None, "숫자만 입력하세요."
        if len(set(s)) != 3:
            return None, "중복 없는 숫자 3자리를 입력하세요."
        return [int(ch) for ch in s], None

    def reset_to_idle(self):
        self.status = 0
        self.btn.setText("게임 시작")
        self.edit.clear()
        self.edit.setFocus()

    def new_game(self):
        # 상태 0이면: 새 게임 시작
        if self.status == 0:
            self.start()
            return

        # 상태 1이면: 입력값 채점
        guess_str = self.edit.text()
        guess, err = self.validate_input(guess_str)
        if err:
            self.view.addItem(f"[입력 오류] {err}")
            self.edit.selectAll()
            self.edit.setFocus()
            return

        self.tries += 1

        strike = sum(1 for i in range(3) if guess[i] == self.answer[i])
        ball = sum(1 for i in range(3) if guess[i] in self.answer and guess[i] != self.answer[i])
        out = 3 - strike - ball

        self.view.addItem(f"{guess_str} -> {strike}S {ball}B {out}Out (시도 {self.tries})")

        self.edit.clear()
        self.edit.setFocus()

        # 3S면 종료 처리
        if strike == 3:
            self.view.addItem("🎉 축하합니다! 정답입니다!")
            self.view.addItem("다시 하려면 '게임 시작'을 눌러주세요.")
            self.reset_to_idle()


if __name__ == "__main__":
    app = QApplication.instance()
    if app is None:
        app = QApplication(sys.argv)
    exam = MyApp()
    sys.exit(app.exec_())
```
