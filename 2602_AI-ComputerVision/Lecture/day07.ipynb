```python
# 회귀(Regression)
X = [1, 2, 3, 4, 5]
y = [30, 50, 70, 90, 100]
#                    110도 진행 예정

# y = W * x + b
# 1차함수, 2차함수, 기울기, 미분, 편미분

# 실제값 = 예측값
# 오차 = 예측값 - 실제값
# 오차1 = W * x1 + b - y1
# 오차2 = W * x2 + b - y2

# 모든 오차 '제곱'의 합
# (오차1)^2 + (오차2)^2 + ... + (오차5)^2
# 최소값을 찾으려면 미분해서 0인 곳 [2차함수 그래프]
# f(x,y) = 2x + 3y + 4
# af(x,y) = 2*dx + 3*dy -> 즉 x, y로 편미분 
```

```python
X = np.array([1, 2, 3, 4, 5])
y = np.array([30, 50, 70, 90, 100])
X_mean = np.mean(X)
y_mean = np.mean(y)
W_분모 = np.sum((X - X_mean) ** 2)
W_분자 = np.sum((X - X_mean) * (y - y_mean))
W = W_분자 / W_분모 # 18
b = y_mean - W * X_mean # 14
print(f"W={W}, b={b}")

def model(xi):
    return W * xi + b
y_pred = model(X)
y_pred
import matplotlib.pyplot as plt
plt.plot(X, y, X, y_pred, marker="o")
plt.show()
```

```python
# 2. 비지도학습(Unsupervised Learing) - 교재 185p
# 정답이 없다
# 군집, 차원 축소, 이상 탐지 정도 진행
# 목표 : 이러한 모델이 있더라 수준

# 1) 군집 알고리즘(Clustering)
from sklearn.datasets import make_blobs # 군집용 데이터 생성
from sklearn.cluster import KMeans # 군집 알고리즘

X, y = make_blobs(n_samples=200, n_features=2, centers=4, cluster_std=1.0, random_state=20)
#                 개수 200,       특징 2개,      센터 4개      군집도 1.0      랜덤 시드

k = 20
kmeans = KMeans(n_clusters=k, random_state=20, n_init="auto")
kmeans.fit(X) #split 안하는 이유는 비지도 학습이라

# 학습 후 2개의 정보 확인
centors = kmeans.cluster_centers_ # 중심점 좌표
labels = kmeans.labels_ # 군집된 결과

# 군집 데이터 시각화
import matplotlib.pyplot as plt
plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.scatter(centors[:, 0], centors[:, 1], c="r")
plt.grid(True)
plt.show()
```

```python
# 차원 축소
# 각 데이터가 64개의 feature를 가지고 있다
# feature 중에 해당 데이터를 잘 나타내는 feature를 위주로 축소
# 조건 : 줄여진 data가 원본 data를 잘 나타내야 함
digits = load_digits()
X = digits.data
y = digits.target
print(f"원본 데이터 차원 : {X.shape}")

# PCA 모델 : 주 성분 분석(Principal Componet Analysis)
# 원본 데이터의 분산이 가장 큰 방향을 새로운 축으로 하여 데이터 투영
# 정사영 -> 위에서 햇빛이 있다 가정하여 그림자가 비추는
#       ㄴ 각 좌표의 거리를 X좌표에 사영하여 X만 사용해도 큰 손실이 없다
from sklearn.decomposition import PCA
pca = PCA(n_components=64)
X_pca = pca.fit_transform(X)
print(f"축소된 데이터 차원 : {X_pca.shape}")

# 축소된 데이터가 원본을 얼마나 반영하는가 - 분산 비율(합으로 사용)
print(f"원본 반영 비율 : {sum(pca.explained_variance_ratio_) * 100: .2f}%")
```

```python
# 이상 탐지
# 연관 규칙 분석(Asscociation Rule Mining)

# 여러분들이 가게 운영
# 빵, 우유가 안 팔림
# 어떤 사람들이 사고, 어떤 사람이 안 사는지 분석
dataset = [
    ['우유', '계란', '빵'],
    ['우유', '기저귀', '맥주', '감자칩'],
    ['계란', '빵', '주스'],
    ['우유', '기저귀', '맥주', '콜라'],
    ['빵', '계란', '우유'],
    ['기저귀', '맥주', '주스'],
    ['빵', '콜라'],
    ['우유', '계란', '빵', '감자칩'],
    ['기저귀', '맥주'],
    ['우유', '빵']
]

# 학습, 분석
# 인코딩이 필요함
# 3, 2.5 -> 수치형 데이터
# "우유", "male", "female" -> 범주형 데이터

# 범주형 데이터 인코딩
# 1) 레이블 인코딩
label_encoded = []
# 전체 리스트 탐색하여
# 처음 보는 데이터면 0부터 1, 2, 3 순으로 숫자를 부여
# 모든 데이터를 바꿔주세요
# 단, 우유 3으로 해도 됨
# 원하는건 같은 품목이면 같은 숫자면 됨
label_dict = {}
label_count = 0
for i in dataset:
    encoded_row = []
    for item in i:
        if item not in label_dict:
            label_dict[item] = label_count
            label_count += 1
        encoded_row.append(label_dict[item])
    label_encoded.append(encoded_row)

print("라벨 사전:", label_dict)
print("인코딩 결과:", label_encoded)
```

```python
# 1) 레이블 인코딩 - 교수님
# 전체 리스트 탐색하여
# 처음 보는 데이터면 0부터 1, 2, 3 순으로 숫자를 부여
# 모든 데이터를 바꾸기 [우유 3으로 해도 됨]

# Python의 dic {key:value}
# {1, 2, 3, 3, 2, 5} -> 중복 없는 자료구조 (Set)
# {1, 2, 3, 5}

# 1) 중복없이
item_list = []
for data in dataset:
    for d in data:
        item_list.append(d)
print(list(set(item_list))) # 중복 제거, but 공간복잡도 엉망
item_list = list(set(item_list))
item_list.sort()

# 2) 아이템 별 Value
dic_item = {}
count = 0
for item in item_list:
    dic_item[item] = count
    count += 1
print(dic_item)

# 3) label 변경
label_encoded = []
for data in dataset:
    temp = []
    for d in data:
        temp.append(dic_item[d])
    label_encoded.append(temp)
print(label_encoded)
# 라이브러리가 있으니까 걱정 ㄴㄴ
# 우유(5)가 과연 계랸(1)보다 5배의 가치가 있느냐??의 문제
```
