```python
# 추상화(Abstraction)
# 객체 설계 시 최소한의 정보(속성, 특성)을 추출하는 행위

# 캡슐화(Encapsulation)
# 사용자들이 접근하는 정보에 대한 제한

# 은행 클래스
class Bank:
    __money = 0 # 전역변수
    def deposit(self, money):
        print(f"입금 : {money}원")
        self.__money += money
    def withdraw(self, money):
        print(f"출금 : {money}원")
        self.__money -= money
    def show_me_the_money(self):
        print(f"잔액 : {self.__money}원")
```

```python
salary_account = Bank()
# 사용자가 접근하면 안되는 멤버에 대해
# 접근 제어(Access Modifier)
# 1) Public* : money -> 누구나 접근 가능!!
# 2) Protected : _money -> 같은 패키지 혹은 상속 관계
# 3) Private* : __money -> 같은 클래스 내에서만
salary_account.money = 1_000_000_000
salary_account.deposit(1000)
salary_account.withdraw(500)

salary_account.show_me_the_money()
```

```python
# 클래스 - 대문자로 시작하는 일종의 규칙
# 객체를 만들기 위한 설계도 용도 이외에
# VO(Value Object) <- 더 많이 사용할 개념

# C언어
# 우리반 연락처를 만들기 위해
# 이름, 나이, MBTI
# 이름 -> 문자열
# 나이 -> 정수형
# MBTI -> 문자열

# 이름 : ㅇㅇㅇ, 나이 : 20(세), MBTI = 0000
class Contact:
    name = ""
    age = 0
    mbti = ""

    def show_contact(self):
        print(f"이름 : {self.name}, 나이 : {self.age}(세), MBTI : {self.mbti}")
```

```python
contact1 = Contact()
contact1.name = "나예호"
contact1.age = 20
contact1.mbti = "ENFJ"
contact1.show_contact()

contact2 = Contact()
contact2.name = "이경환"
contact2.age = 20
contact2.mbti = "INFP"
contact2.show_contact()

list_contact = [contact1, contact2]
total_age = 0
for contact in list_contact:
    total_age += contact.age

print(f"학생의 평균 나이 : {total_age / len(list_contact) : .0f}세")
```

```python
# 포켓몬 VO 만들어 보자 -> 명세
# 클래스 이름 : Pokemon
# 멤버변수(field) : name, p_type, height, weight
# 멤버함수(function) : show_info()
# 이름 : 꼬부기, 타입 : 물, 키 : 0.5m, 몸무게 : 9.0kg
# 혹시 구성요소를 빼먹었을 때 쓰는 기능
class Pokemon:
    # __init__이 가장 먼저 실행되어 name 따로 정의 X
    # init안에 있는 멤버 변수는
    # 대체적으로 private(__) 속성으로 선언!!
    def __init__(self, name, p_type, height, weight):
        self.__name = name
        self.__p_type = p_type
        self.__height = height
        self.__weight = weight

    # 키만 변경할 수 있는 기능(함수)를 만들어주자
    # Setter 함수
    def set_height(self, height):
        self.__height = height
        
    # 몸무게를 반환하는 함수
    # Getter 함수
    def get_weight(self):
        return self.__weight

    # 부르면 호출되는 내장함수 __str__
    # 반드시 리턴이 있어야됨
    def __str__(self):
        return f"이름 : {self.__name}, 타입 : {self.__p_type}, 키 : {self.__height}m, 몸무게 : {self.__weight}kg"
```

```python
# Pokemon() <- 생성자
# 실제로 만들때 def __init(self): pass 가 실제로 올라감
num007 = Pokemon("꼬부기", "물", 0.5, 9.0)
print(num007)
num004 = Pokemon("파이리", "불", 0.6, 8.5)
print(num004)
# 꼬부기의 키를 0.6으로 조정
# private으로 설정했으니 외부에서 접근 X
num007.height = 0.6
num007.name = "이상해씨"
num007.__name = "이상해씨"
print(num007)
num007.set_height(0.6)
print(num007)
```

```python
# 음악 관련 VO -> 명세
# Music 클래스 설계
# field : title, artist
# __init__() 활용해서 설계!!
# 모든 field는 private
# 모든 field에 대해 Getter, Setter 정의
# __str__() 함수는 다음을 반환한다
# 제목 : Good Goodbye, 가수 : 화사

class Music:
    def __init__(self, title, artist):
        self.__title = title
        self.__artist = artist

    def set_title(self, title):
        self.__title = title
        
    def get_title(self):
        return self.__title

    def set_artist(self, artist):
        self.__artist = artist
        
    def get_artist(self):
        return self.__artist

    def __str__(self):
        return f"제목 : {self.__title}, 가수 : {self.__artist}"
```

```python
m1 = Music("Good Goodbye", "화사")
m2 = Music("Golden", "HUNTR/X")
m3 = Music("그대 작은 나의 세상이 되어", "WOODZ")
print(m1)
print(m3)
m3.set_artist("카더가든")
print(m3)
```

```python
# 상속(Inheritance)

# 호랑이, 코끼리, 아르마딜로, 랫서판다
# 위 4 객체의 공통적인 특징 추출 -> 포유류
# 상속을 하는 클래스 : 부모 클래스, super class, 상위 클래스
# 상속을 받는 클래스 : 자식 클래스, sub class, 하위 클래스

# 포유류 클래스를 만들어보자
class Mammal:

    def __init__(self, weapon, height, weight):
        self.__weapon = weapon
        self.__height = height
        self.__weight = weight

    def attack(self):
        print(f"{self.__weapon}(으)로 공격!!")

    def feed(self):
        print("젖을 먹인다")
```

```python
# Mammal을 상속 받아 클래스를 정의하겠다
class Red_Panda(Mammal):
    def cute(self):
        print("귀여움")

# 다형성(Plymorphism)
# 부모 클래스의 함수를 자식 클래스에서 재정의
# -> Overriding
class Whale(Mammal):
    # 고래는 weapon,height, weight 이외 feature 추가
    def __init__(self, feature, weapon, height, weight):
        super().__init__(weapon, height, weight)
        self.__feature = feature

    def swim(self):
        print("헤엄~ 헤엄~")

    def attack(self):
        print("몸통 박치기")
```

```python
푸바오 = Red_Panda("발톱", 0.5, 6.0)
푸바오.cute()
푸바오.feed()

# 다형성(Polymorphism)
gorae = Whale("지느러미", "큰 입", 100, 2000)
gorae.swim()
gorae.attack()
gorae.feed()
```

```python
# 객체 지향 프로그래밍(OOP)
# 독립적인 객체(Data, Logic)를 만들어
# 메세지를 통해 프로그램을 설계하는 패러다임

# 객체지향 프로그래밍 특징 4가지
# 추상화(Abstraction)
# 객체의 공통적인 특징을 추출하여 클래스를 설계
# 코끼리 -> 상아, 코가 길어요, 덩치가 크다
# 코끼리, 호랑이, 아르마딜로, 랫서 판다의
# 공통적인 특징인 포유률 자체를 추출하는 행위도 추상화
# 포유류, 조류, 어류 -> 생물

# 캡슐화(Encapsulation)
# 사용자가 굳이 몰라도 되는 정보(멤버변수, 멤버함수)를
# 접근 제어자(Access Modifier)를 통해 제어하는 행위
# 개발자가 의도한 기능만 공개하는 행위
#  ㄴ 정보 은닉: 알면 안되는 정보(보안 이슈)를 감추는 행위

# 상속(Inheritance)
# 이미 만들어진 검증된 클래스를 물려 받아 새로운 클래스를 정의
# 물려 주는 클래스: 부모 클래스, Super Class, 상위 클래스
# 물려 받는 클래스: 자식 클래스, Sub Class, 하위 클래스
# 자식 클래스 공통 형질 추상화 -> 부모 클래스 정의

# 다형성(Polymorphism)
# 다양한 형태를 나타낼 수 있는 능력
# 함수 하나 호출 -> 여러 결과가 나온다
# 상속 받은 자식 클래스에서 Overriding(함수 재정의)

# 클래스의 다른 용도
# VO(Value Object): 사용자 정의 객체
# 사용자가 원하는 자료형을 만든다 (마치 C언어의 구조체처럼)
# __init__() -> 생성자 함수
# __str__() -> 인스턴스명 호출 시 반환되는 함수
# Getter, Setter
# DAO(Data Access Object) & DTO(Data Transfer Object)
```

```python
# 디지몬 클래스 명세
# 클래스 이름: Digimon
# 멤버변수: name, tier(등급), special
# 멤버함수(function): attack, s_attack

# 아구몬, 파피몬
# 배틀, 둘다 hp는 100으로 출발
# attack이 될 확률은 90%, s_attack은 10%
# 만약, s_attack이 발동하면 "쁘띠 파이어!!"
# 상대방 hp 50 감소, 그냥 공격은 10 감소
# 서로 한 번씩 공격하면서
# 이긴 디지몬 이름 불러주기
```

```python
class Digimon:
    def __init__(self, name, tier, special):
        self.__name = name
        self.__tire = tier
        self.__special = special

    def get_name(self):
        return self.__name
    
    def attack(self):
        print("공격!")

    def s_attack(self):
        print(f"{self.__special}(으)로 공격!!")

d1 = Digimon("아구몬", "성장기", "베이비 플레임")
d2 = Digimon("파피몬", "성장기", "쁘띠 파이어")
# 랜덤 90퍼 확률을 어떻게??
import random

d1_hp, d2_hp = 100, 100
while True:
    dice1 = random.randint(1, 10)
    print(f"{d1.get_name()}의 턴 :", end=" ")
    if dice1 == 7:
        d1.s_attack()
        d2_hp -= 50
    else:
        d1.attack()
        d2_hp -= 10
    if d2_hp <= 0:
        print(f"{d1.get_name()}의 승리!!")
        break
        
    dice2 = random.randint(1, 10)
    print(f"{d2.get_name()}의 턴 :", end=" ")
    if dice2 == 7:
        d2.s_attack()
        d1_hp -= 50
    else:
        d2.attack()
        d1_hp -= 10
    if d2_hp <= 0:
        print(f"{d2.get_name()}의 승리!!")
        break
    
```

```python
# 자료구조
# 1) 선형 자료구조
#   1. 연결 리스트(Linked List) : Node(Data & Address) -> 1방향
#                            ㄴ2방향이상이면 Tree라고 부름
# 2) 비선형 자료구조
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
# Node(10), Node(20) 등으로 호출 예정
# Node(10) -> Node(20) -> Node(30) 이면 노드
# Node(10) -> Node(20) or Node(30) 식이면 트리
# Node(10) 위에 20, 위에 30 이면 스택
# 10이 안쪽, 그 다음이 20, 바깥쪽이 30 이면 큐
```

```python
# 1) 연결 리스트(Linked List)
class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        new_node = Node(data)

    # 만약, 아무런 노드가 없다면
    # 새로 들어온 노드를 head로 임명!!
        if self.head is None:
            self.head = new_node
        else:
        # 맨 끝 노드까지 이동 -> 거기서 연결
        # 주의) 연결 리스트에서 head는 항상 철 노드를 가리켜야 한다.
        #  -> self.head 에 절대 대입 금지
            current = self.head
            while current.next is not None:
                current = current.next
            current.next = new_node

    # 연결된 링크 Print
    def display(self):
        # 10 -> 20 -> 30 으로 출력되길 원함
        if self.head is None:
            print("연결 리스트가 비어 있습니다.")
            return

        current = self.head
        while current: #current.data도 되지만 마지막 공백 오류
            print(f"{current.data} -> ", end="") # 10 출력
            current = current.next
        print("None")

    # 탐색
    def search(self, target):
        # target이 전 노드를 탐색하면서 있다면!!
        # index를 반환
        # 찾았다!! 40은 3번 index에 있습니다.!!
        # 없다면 -> 찾을 수 없습니다.
        index = 0
        current = self.head

        while current:
            if target == current.data:
                print(f"찾았다!! {target}은 {index}번 index에 있습니다!!")
                return
            current = current.next
            index += 1
        print("찾을 수 없습니다.")
        return

    # 수정
    def update(self, target, new_data):
        current = self.head
        while current:
            if target == current.data:
                print("값 변경 완료")
                current.data = new_data
                return
            current = current.next
        print("항목을 찾을 수 없습니다.")
        return

    # 삭제
    def delete(self, target):
        current = self.head
        prev = None

        # if self.head.data == target:
            # self.head = current.next
            # return
        # while current.next:
            # if current.next.data == target:
                # current.next = current.next.next
                # return
            # current = current.next
        
        while current:
            if target == current.data:
                print("값 제거 완료")
                if prev is None:
                    self.head = current.next
                else:
                    prev.next = current.next
                return

            prev = current
            current = current.next

        print("항목을 찾을 수 없습니다.")
        return
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        
class Stack:
    def __init__(self):
        self.top = None

    def is_empty(self):
        return self.top is None

    def push(self, data):
        new_node = Node(data)
        if self.is_empty():
            self.top = new_node
        else:
            current = new_node
            current.next = self.top
            self.top = current
            

    def peek(self):
        if self.is_empty():
            return None
        return self.top.data

    def pop(self):
        if self.is_empty():
            return None
        current = self.top
        self.top = self.top.next
        return current.data
```
