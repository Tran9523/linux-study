```python
# live webcam code
import cv2
import time
from ultralytics import YOLO

cap = cv2.VideoCapture(0)
# width, height = resolution
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 640)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 480)

model = YOLO("yolov8n.pt") # YOLO add
# ncnn
model.export(format="ncnn", imgsz=320, half=True)
# model.export(format="tflite", int8=True)

if not cap.isOpened():
    exit()
try:
    while True:
        start = time.time()
        ret, frame = cap.read()
        if not ret:
            break

        detect = model(frame)
        
        end = time.time()
        fps = 1 / (end - start) # frame per second
        cv2.putText(frame, f"fps:{fps:.2f}", (20, 50), cv2.FONT_HERSHEY_PLAIN, 3, (0,0,255), 2)
        cv2.imshow("frame", detect[0].plot())
        # cv2.imshow("frame", frame)
        
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break

    cv2.destroyAllWindows()
    cv2.waitKey(1)
finally:
    cap.release()
```
