```python
# 강아지 사진
# "dog.jpg" 를 YOLOv8로
from ultralytics import YOLO
import cv2
# 없으면 다운로드, 있으면 그대로 진행
model = YOLO("yolov8s.pt") # YOLOv8의 s모델을 쓰자
frame = cv2.imread("dog.jpg")

detect = model(frame)
# 1이 검측된 물체 개수 라는걸 알 수 있음
# YOLOv8은 plot()함수를 통해, box, class_name, confidence까지 잘 나오게 함수화
cv2.imshow("here", detect[0].plot())
# cv2.imshow("frame",frame) # 얘 날려도 됨
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.waitKey(1)
```

```python
# 실시칸 웹캠을 통해 YOLOv8 모델 돌려보기
from ultralytics import YOLO
import cv2
# 없으면 다운로드, 있으면 그대로 진행
model = YOLO("yolov8s.pt") # YOLOv8의 s모델을 쓰자

cap = cv2.VideoCapture(0)
if not cap.isOpened():
    exit()
try:
    while True:
        ret, frame = cap.read()
        if not ret:
            print("프레임을 읽을 수 없습니다")
            break
        detect = model(frame)
        cv2.imshow("frame", detect[0].plot())
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break
finally:
    cap.release()
    cv2.destroyAllWindows()
    cv2.waitKey(1)
```
