```python
# 예외처리
# 실행 시 Error Class
# try ~ except : 에러가 날거 같은 구간을 알아야됨

try:
    num1 = int(input("첫 번째 정수 입력 >>"))
    num2 = int(input("두 번째 정수 입력 >>"))

    if num1 % num2 == 0:
        print(f"{num2}는 {num1}의 약수입니다.")
    else:
        print(f"{num2}는 {num1}의 약수가 아닙니다.")
except ValueError:
    print("정수를 입력해주세요.")
except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")
finally: # 예외 발생 상관 없이 실행되는 부분
    print("프로그램 종료")
```

```python
# temp.txt 를 읽었을 때 발생하는 에러 클래스 확인 후
# 해당 파일이 있다면 -> read()를 통해 모든 내용 출력
# 없다면 -> "해당 파일이 존재하지 않습니다." 출력!!
# 예외 발생 상관 없이 "프로그램 종료" 출력
try:
    with open("temp.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("해당 파일이 존재하지 않습니다.")
finally:
    print("프로그램 종료")
```

```python
with open("salary.csv", "r", encoding="euc-kr") as f:
    reader = csv.DictReader(f)
    for row in reader:
        if "10%" in row["구분"] :
            earned_income = int(row["근로소득금액"]) / int(row["인원"])
            print(f"10% 구간 근로소득 : {earned_income:.2f}억 원")
```

```python
with open("salary.csv", "r", encoding="euc-kr") as f:
    reader = csv.DictReader(f)
    total_popular = 0
    total_salary = 0
    for row in reader:
        total_popular += int(row["인원"])
        total_salary += int(row["근로소득금액"])
print(f"평균 근로소득 : {total_salary / total_popular * 10000 : .2f}만원")
```

```python
with open("salary.csv", "r", encoding="euc-kr") as f:
    reader = csv.DictReader(f)
    people = 0
    salary = 0
    count = 0
    search = "40%"
    for row in reader:
        people += int(row["인원"])
        salary += int(row["근로소득금액"])
        count += 1
        if search in row["구분"]:
            earned_income = int(row["근로소득금액"]) / int(row["인원"]) * 10000
            final_tax = int(row["결정세액"]) / int(row["인원"]) * 10000
            print(f"상위 {search} 구간의 근로소득금액: {earned_income:.2f}만원")
            print(f"상위 {search} 구간의 결졍세액: {final_tax:.2f}만원")

    print(f"평균 근로소득 : {salary / people * 10000 : .2f}만원")
```

```python
df_poke = pd.read_csv("pokemon.csv")
# df_poke.info()
df_poke.describe() # 수치데이터를 확인
df_poke.head()
# DataFrame 기본 인덱싱 및 슬라이싱
# 인덱싱
# 데이터프레임[컬럼명] -> 1차원 시리즈
df_poke["Name"]
# 데이터프레임[["컬럼명1", "컬렁명2"]] -> 2차원 데이터 프레임
df_poke[["Name", "Total"]]
```

```python
# 조건에 맞는 데이터 추출
# loc (location)
# 데이터프레임.loc[행 인덱싱 값, 열 인덱싱 값]
df_poke.loc[0] # 0번 인덱스의 data -> 1차원 시리즈
df_poke.loc[0, "Total"] # 0번 인덱스의 Total 컬럼 data
df_poke.loc[0, "Name"] # 0번 인덱스의 Name 컬럼 data

# 데이터프레임.loc[행 슬라이싱]
df_poke.loc[0 : 2] # 0번 인덱스 ~ 2번 인덱스

# 데이터프레임.loc[행 슬라이싱, 열 슬라이싱]
df_poke.loc[0 : 2, "Name":"Total"] # 0 ~ 2번 인덱스, Name ~ Total 컬럼
```

```python
# Boolean Indexing
# 포켓몬 Total 이 400이상인 데이터들만 확인!!
df_poke["Total"] >= 400

# 데이터프레임[행 인덱싱] <- 여기에 series를 넣을꺼임
df_poke[df_poke["Total"] >= 400]

# 만약 그 조건을 만족하는 1세대 포켓몬만 [조건이 & 하나만]
df_poke.loc[(df_poke["Total"] >= 400) & (df_poke["Generation"] == 1)]
```

```python
# Type1이 Fire인 포켓몬의 개수 = 76
df_poke.loc[df_poke["Type1"] == "Fire"]
# Type1이 Water인 포켓몬 중 Total 500이상인 포켓몬의 개수 = 47
# 개수는 출력 시 나오는 rows를 확인하자!!
df_poke.loc[(df_poke["Type1"] == "Water") & (df_poke["Total"] >= 500)].count()
```

```python
# 데이터프레임 인덱싱 및 슬라이싱
# loc는 정확히 일치했을때, iloc는 앞에서부터 차례대로
# iloc (Integer Location) **가장 많이 씀
# 데이터프레임.iloc[행 인덱스 정수]
df_poke.iloc[0]
# 데이터프레임.iloc[행 인덱스 정수, 열 인덱스 정수]
df_poke.iloc[0, 1]

# 데이터프레임.iloc[행 인덱스 정수 슬라이싱]
# iloc -> 마지막 범위 포함 X
df_poke.iloc[0 : 3]

# 데이터프레임.iloc[행 인덱스 정수 슬라이싱, 열 인덱스 정수 슬라이싱]
df_poke.iloc[0 : 3, 1 : 6]
```
