```python
# 최소 동전의 문제
# 8700원
money = int(input("원 입력하세요 :"))
# 천원 8개, 백원 7개 -> 15개
# 오천원 1개, 천원 3개, 백원 7개 -> 11개
# 오천원 1개, 천원 3개, 오백원 1개, 백원 2개 -> 7개
money_5000 = (money // 5000)
money_last1 = money % 5000
money_1000 = (money_last1 // 1000)
money_last2 = money % 1000
money_500 = (money_last2 // 500)
money_last3 = money % 500
money_100 = (money_last3 // 100)
print(f"오천원 : {money_5000}장")
print(f"천원 : {money_1000}장")
print(f"오백원 : {money_500}개")
print(f"백원 : {money_100}개")
# money_1000 = (money - (money_5000 * 5000)) // 1000
# money_500 = money % 5000 % 1000 // 500
# money_100 = money % 500 // 100
```

```python
time = int(input("시간 입력 >> "))
hour = time // (60 * 60)
minute = time % (60 * 60) // 60
second = time % 60
print("{}시간 {}분 {}초".format(hour, minute, second))

# 코드를 작성할 때,
# 결과는 동일하되
# 가독성을 높이거나, 실행속도를 높이거나
# Refactoring
# Design = 보기 좋게 변수를 모아두거나 분류하는 등등
```

```python
# 3) 대입 연산자
# 우항 '값'이 좌항 '변수'에 할당
salary = 4_000_000
# 월급 10% 상향!!
# 복합대입연산자
# =과 함께 산술연산자, 지수연산자 가능
# +=, -=, *=, /=, //=, %= 등
salary *= 1 + (0.1)
print(f"월급 : {salary}원")
```

```python
# 4) 비교 연산자
# >, >=, <, <=, ==, !=
# 비교 연산자의 결과는 항상 논리값이다!!
print(3 > 5)
print(3 >= 5)
print(3 < 5)
print(3 <= 5)
print(3 == 5)
print(3 != 5)
```

```python
# 5) 논리 연산자 (not, and, or)
# not : 논리 결과를 부정(T -> F, F -> T)
# 반에 불이 꺼져 있습니다
# 만약, 반에 불이 꺼져 잇으면 켜고
# 켜져 있으면 꺼주세요
isLight = False
isLight = not isLight

# and (논리곱)
# 논리값 and 논리값
# 둘다 Trun여야만 Trun가 나온다!!
# 1 x 1 = 1 (T x T -> T)
# 1 x 0 = 0 (T x F -> F)
# 0 x 1 = 0 (F x T -> F)
# 0 x 0 = 0 (F x F -> F)
# 만약, and 뒷쪽에 반드시 실행되어야 하는 코드가 있을 경우
# 주의!! Shortcut circuit
# 앞이 False라면 뒷쪽은 Dead Code(실행 X)
print((3 > 5) and (10 > 5))

# or (논리합)
# 논리값 or 논리값
# 둘다 False일 때만, False가 나온다
# 1 x 1 = 1 (T x T -> T)
# 1 x 0 = 1 (T x F -> T)
# 0 x 1 = 1 (F x T -> T)
# 0 x 0 = 0 (F x F -> F)
# 만약, or 뒷쪽에 반드시 실행되어야 하는 코드가 있을 경우
# 주의!! Shortcut circuit
# 앞이 True라면 뒷쪽은 Dead Code(실행 X)
```

```python
# 조건(삼항) 연산자

# 항 : 연산의 영향을 받는 것
# ㄴ a + b

# a if 조건식 else b
# 조건식 : 논리값(Trun, False)
score = 90
grade = "A" if score >= 90 else "F"
print(grade)
```

```python
#2개의 정수를 입력받아 큰수에서 작은수 빼기
num1 = int(input("첫 번째 정수 입력 >>"))
num2 = int(input("두 번째 정수 입력 >>"))
result = (num1 - num2) if (num1 > num2) else (num2 - num1)
print(f"두 수의 차 : {result}")
print(f"두 수의 차 : {num1 - num2 if num1 > num2 else num2 - num1}")
```
