```python
# 사실...Python에서 List를 쓰면
# 배열을 구현 안해도 된다
class Array:
    def __init__(self):
        self.data = []

    def append(self,data):
        self.data.append(data)

    def get(self, index):
        return self.data[index]

    def update(self, index, data):
        # 수정 Logic 구현!!
        self.data[index] = data

    def pop(self):
        # 가장 마지막 요소 삭제
        # self.data[len(self.data)-1] = None #잘 안되네
        self.data.pop()
        
    def remove(self, index):
        # 특정 인덱스 요소 삭제
        # self.data[index] = None #얘도 똑같음
        del self.data[index]

    def __str__(self):
        # for i in range(len(self.data)):
            # print(f"{self.data[i]}", end=" ")
        # print()
        return str(self.data) #모든 요소 출력 [10, 20, 30]
```

```python
arr = Array()
arr.append(10)
arr.append(20)
arr.append(30)
arr.append(40)
arr.append(50)
print(arr)
print(arr.get(1)) # 1번 인덱스의 값을 반환
arr.update(1, 50)
print(arr.get(1))
arr.pop()
print(arr)
arr.remove(2)
print(arr)
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

```python
# Queue(큐)
# First In First Out

class Queue:
    def __init__(self):
        self.head = None
        self.tail = None
        self.cnt = 0 # size를 구현하기 위한 개수 정보

    # push(10)
    # 1) 새로운 노드를 new_node
    # head와 tail이 해당 노드를 가리키게!!
    def push(self, data):
        new_node = Node(data)
        if self.head is None:
            self.head = self.tail = new_node
        else:
            current = self.tail
            current.next = new_node
            self.tail = new_node
            # current = new_node
            # self.tail.next = current
            # self.tail = current
        self.cnt += 1
            
    def empty(self):
        return self.head is None

        # pop(dequene 라고 부름)
    def pop(self):
        if self.empty():
            self.cnt = 0
            return -1
            
        current = self.head
        self.head = self.head.next #current.next 도 동일
        self.cnt -= 1
        if self.empty():
            self.tail = None
        return current.data

    def size(self):
        # 큐에 들어있는 정수 개수 반환!!
        # 비어 있으면 -> 0
        # 우리에게 주어진 정보는 head, tail
            return self.cnt

    def front(self):
        if self.empty():
            return -1
        return self.head.data
        # return -1 if self.empty() else self.head.data
    
    def back(self):
        if self.empty():
            return -1
        return self.tail.data
        
# push X: 정수 X를 큐에 넣는 연산이다.
# pop: 큐에서 가장 앞에 있는 정수를 빼고, 그 수를 출력한다. 만약 큐에 들어있는 정수가 없는 경우에는 -1을 출력한다.
# size: 큐에 들어있는 정수의 개수를 출력한다.
# empty: 큐가 비어있으면 1, 아니면 0을 출력한다.
# front: 큐의 가장 앞에 있는 정수를 출력한다. 만약 큐에 들어있는 정수가 없는 경우에는 -1을 출력한다.
# back: 큐의 가장 뒤에 있는 정수를 출력한다. 만약 큐에 들어있는 정수가 없는 경우에는 -1을 출력한다.
```

```python
q = Queue()
q.push(10)
q.push(20)
q.push(30)
print(q.tail.data)
print(q.head.data)
print(q.head.next.data)
print(q.head.next.next.data)
print(q.size())
q.pop()
print(q.head.data)
print(q.head.next.data)
print(q.size())
print(q.front())
print(q.back())
```

```python
# 변외로 Queue를 Python의 List를 활용하여 구현해보자
class CustomQueue:
    def __init__(self):
        self.data = []

    def push(self, data):
        self.data.append(data)

    def empty(self):
        return len(self.data) == 0
        
    def pop(self):
        if self.empty():
            return -1
        else:
            temp = self.data[0]
            del self.data[0]
            return temp

    def size(self):
        return len(self.data)

    def front(self):
        if self.empty():
            return -1
        return self.data[0]

    def back(self):
        if self.empty():
            return -1
        return self.data[-1] #가장 뒤에 있는 요소
```

```python
cq = CustomQueue()
cq.push(10)
cq.push(20)
cq.push(30)
print(cq.pop())
cq.push(40)
print(cq.size())
print(cq.front())
print(cq.back())
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class Queue:
    def __init__(self):
        self.head = None
        self.tail = None
        self.cnt = 0

    def push(self, data):
        new_node = Node(data)
        if self.head is None:
            self.head = self.tail = new_node
        else:
            current = new_node
            self.tail.next = current
            self.tail = current
        self.cnt += 1

    def empty(self):
        return self.head is None

    def pop(self):
        if self.empty():
            self.tail = None
            self.cnt = 0
            return -1        
        current = self.head
        self.head = current.next
        self.cnt -= 1
        return current.data

    def size(self):
        return self.cnt

N = int(input())
q = Queue()
for i in range(1, N + 1):
    q.push(i)

while True:
    if q.size() == 1:
        break
    else:
        print(q.pop(), end=" ")

    if q.size() == 1:
        break
    else:
        q.push(q.pop())

print(q.pop())
```

```python
# Dictionary
# 자료 구조 중 하나

# Python -> C
# 이럴때 언어 변환이 필요함

# CSV(Comma Separated Value)
# 이름, 나이, MBTI   ->   이런 구조를 Header라고 함
# 나예호, 20, ENFJ
# 수지, 30, INFP
# -> 공백 포함 42byte
# 장점 : 대량 데이터를 다루기 용이함
# 단점 : 데이터의 정보 확인 어려움

# XML(eXtensive Markup Language)
# <이름>"나예호"</이름><나이>20</나이><MBTI>"ENFJ"</MBTI>
# <이름>"수지"</이름><나이>30</나이><MBTI>"INFP"</MBTI>
# -> 공백 포함 109byte
# 장점 : 메타데이터 포함 -> 정보 확인 쉽다
# 단점 : 비용이 많이 든다

# JSON(JavaScript Object Notation)
# {"이름":"나예호", "나이":20, "MBTI":"ENFJ"},
# {"이름":"수지", "나이":30, "MBTI":"INFP"}
# -> 공백 포함 86byte
# 장점 : CSV 대비 정보 확인 쉽다, XML 대비 비용 적게 든다
# 단점 : CSV 대비 비용 많이 든다, XML 대비 정보 확인 어렵다
```

```python
# Python에서 JSON을 Dictionary라고 부른다
dic_yh = {"이름":"나예호", "나이":20, "MBTI":"ENFJ"}
dic_sz = {"이름":"수지", "나이":30, "MBTI":"INFP"}
print(dic_yh)
print(dic_sz)

# Dictionary 값 확인
# :을 기준으로 key-value 쌍으로 구성
# 딕셔너리명[키값]
print(dic_yh["나이"]) #없는 키 가져오면 Error
print(dic_yh.get("나이")) #없는 키 가져오면 None

# Dictionary 값 추가
# 딕셔너리명[키값] = 밸류값
dic_yh["혈액형"] = "AB"
print(dic_yh)

# Dictionary 값 수정
# 딕셔너리명[키값] = 밸류값 #즉 없으면 추가, 있으면 수정
dic_yh["나이"] += 1
print(dic_yh)
```

```python
dic_list = [dic_yh, dic_sz]
# 나이의 총합 / 키 밸류를 자주 써야하니 안전하게 복붙으로
age_total = 0
for dic in dic_list:
    age_total += dic["나이"]

print(age_total)
print(dic_list)
```
